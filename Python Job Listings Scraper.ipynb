```python
import requests
from bs4 import BeautifulSoup
import csv
```

```python
url = 'https://realpython.github.io/fake-jobs/'
responce = requests.get(url)
print(responce.status_code)
```

### code 200 means the request was successful

```python
soup = BeautifulSoup(responce.text, 'html.parser')
```

```python
print(soup.title.text)
```

```python
print(soup.prettify()[:2000])
#HTML Structure
```

```python
jobs = soup.find_all('div', class_ = 'card-content')
print("Total JOB postings:", len(jobs))
```

### Extracting the job details

```python
job = jobs[0]
title = job.find('h2', class_='title').text.strip()
company = job.find('h3', class_='company').text.strip()
location = job.find('p', class_='location').text.strip()

print("Job Title:", title)
print("Company:", company)
print("Location:", location)
```

```python
link = job.find('a', string='Apply')
job_url = link['href'] if link else None
```

```python
print(job_url)
```

```python
job_list = []
for job in jobs:
    title = job.find('h2', class_='title')
    company = job.find('h3', class_='company')
    location = job.find('p', class_='location')
    link = job.find('a', string='Apply')
    
    job_data = {'Job Title': title.text.strip() if title else None,
                'Company': company.text.strip() if company else None,
                'Location': location.text.strip() if location else None,
                'Job URL': link['href'] if link else None  }
    job_list.append(job_data)
    
print("Total jobs extracted:", len(job_list))
#Extract all job listings
```

## Converting into DataFrame

```python
import pandas as pd
df = pd.DataFrame(job_list)
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

## Exporting to CSV

```python
with open('job_listings.csv', 'w', newline='', encoding='utf-8') as file:
    writer = csv.DictWriter(file, fieldnames=['Job Title', 'Company', 'Location', 'Job URL'])
    writer.writeheader()
    writer.writerows(job_list)
```
